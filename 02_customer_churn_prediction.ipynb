```python
import pandas as pd
import numpy as np

from pathlib import Path

print("Libraries imported successfully!")
```

```python
# Define the project directory
BASE_DIR = Path.cwd()

# Load the cleaned sales dataset
DATA_PATH = BASE_DIR / "data" / "cleaned_sales.csv"

df = pd.read_csv(DATA_PATH)

# Convert columns to their correct data types
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"], errors="coerce")
df["CustomerID"] = df["CustomerID"].astype("Int64")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python
print("Total records:", len(df))
print("Unique customers:", df["CustomerID"].nunique())
print("Date range:")
print("Start:", df["InvoiceDate"].min())
print("End:", df["InvoiceDate"].max())

df[["CustomerID", "InvoiceNo", "InvoiceDate", "Quantity", "UnitPrice"]].head()
```
