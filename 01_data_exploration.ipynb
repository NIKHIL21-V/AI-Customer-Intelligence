````python
df_sales["InvoiceDate"] = pd.to_datetime(
    df_sales["InvoiceDate"],
    errors="coerce"
)

print("Invoice dates converted successfully!")
print(df_sales["InvoiceDate"].dtype)
````

````python
best_month = monthly_sales.loc[
    monthly_sales["Revenue"].idxmax()
]

print("Best-performing month:", best_month["YearMonth"])
print("Revenue:", round(best_month["Revenue"], 2))
````

````python
# Identify cancelled invoices
cancelled_mask = (
    df_customer["InvoiceNo"]
    .astype(str)
    .str.startswith("C")
)

cancelled_orders = df_customer[cancelled_mask]

print("Cancelled transaction lines:", len(cancelled_orders))
print("Unique cancelled invoices:", cancelled_orders["InvoiceNo"].nunique())

display(cancelled_orders.head())
````

````python
negative_quantity = df_customer[df_customer["Quantity"] < 0]

print("Transaction lines with negative quantities:", len(negative_quantity))

display(
    negative_quantity[
        ["InvoiceNo", "StockCode", "Description", "Quantity", "UnitPrice"]
    ].head()
)
````

````python
completed_mask = (
    ~df_customer["InvoiceNo"].astype(str).str.startswith("C")
    & (df_customer["Quantity"] > 0)
    & (df_customer["UnitPrice"] > 0)
)

df_completed_sales = df_customer.loc[completed_mask].copy()

df_completed_sales["Revenue"] = (
    df_completed_sales["Quantity"] * df_completed_sales["UnitPrice"]
)

print("Original customer transaction lines:", len(df_customer))
print("Completed-sales transaction lines:", len(df_completed_sales))
print("Preliminary completed-sales revenue:",
      round(df_completed_sales["Revenue"].sum(), 2))
````

````python
kpis = {
    "Gross Sales Before Returns": df_completed_sales["Revenue"].sum(),
    "Total Orders": df_completed_sales["InvoiceNo"].nunique(),
    "Unique Customers": df_completed_sales["CustomerID"].nunique(),
    "Unique Products": df_completed_sales["StockCode"].nunique()
}

for metric, value in kpis.items():
    if "Sales" in metric:
        print(f"{metric}: £{value:,.2f}")
    else:
        print(f"{metric}: {value:,}")
````

````python
# Find the top 10 products by quantity sold

top_products = (
    df_completed_sales
    .groupby("Description")
    .agg(
        Total_Quantity=("Quantity", "sum"),
        Total_Revenue=("Revenue", "sum")
    )
    .sort_values("Total_Quantity", ascending=False)
    .head(10)
    .reset_index()
)

display(top_products)
````

````python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

plt.barh(
    top_products["Description"],
    top_products["Total_Quantity"]
)

plt.xlabel("Total Quantity Sold")
plt.ylabel("Product")
plt.title("Top 10 Best-Selling Products")

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
````

````python
# Calculate revenue generated by each customer

customer_revenue = (
    df_completed_sales
    .groupby("CustomerID")
    .agg(
        Total_Revenue=("Revenue", "sum"),
        Total_Orders=("InvoiceNo", "nunique"),
        Total_Quantity=("Quantity", "sum")
    )
    .sort_values("Total_Revenue", ascending=False)
    .reset_index()
)

print("Top 10 Customers by Revenue")
display(customer_revenue.head(10))
````

````python
# Visualize the top 10 customers by revenue

top_customers = customer_revenue.head(10).copy()

top_customers["CustomerID"] = (
    top_customers["CustomerID"].astype(str)
)

plt.figure(figsize=(10, 6))

plt.barh(
    top_customers["CustomerID"],
    top_customers["Total_Revenue"]
)

plt.xlabel("Total Revenue (£)")
plt.ylabel("Customer ID")
plt.title("Top 10 Customers by Revenue")

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
````

````python
# Step 20: RFM Customer Analysis

import pandas as pd

# Ensure InvoiceDate is in datetime format
df_completed_sales["InvoiceDate"] = pd.to_datetime(
    df_completed_sales["InvoiceDate"]
)

# Set analysis date to one day after the last transaction
analysis_date = (
    df_completed_sales["InvoiceDate"].max()
    + pd.Timedelta(days=1)
)

# Calculate RFM metrics
rfm = (
    df_completed_sales
    .groupby("CustomerID")
    .agg(
        Recency=(
            "InvoiceDate",
            lambda dates: (analysis_date - dates.max()).days
        ),
        Frequency=("InvoiceNo", "nunique"),
        Monetary=("Revenue", "sum")
    )
    .reset_index()
)

# Display the first 10 customers
display(rfm.head(10))

print("Total customers analyzed:", len(rfm))
````

````python
# Summarize customer purchasing behavior

display(
    rfm[["Recency", "Frequency", "Monetary"]].describe().round(2)
)
````

````python
# Step 21: Assign RFM scores

rfm["R_Score"] = pd.qcut(
    rfm["Recency"].rank(method="first"),
    q=4,
    labels=[4, 3, 2, 1]
).astype(int)

rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    q=4,
    labels=[1, 2, 3, 4]
).astype(int)

rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    q=4,
    labels=[1, 2, 3, 4]
).astype(int)

# Combine the three scores
rfm["RFM_Score"] = (
    rfm["R_Score"].astype(str)
    + rfm["F_Score"].astype(str)
    + rfm["M_Score"].astype(str)
)

display(rfm.head(10))
````

````python
# Assign business-friendly customer segments

def classify_customer(row):
    if row["R_Score"] >= 3 and row["F_Score"] >= 3 and row["M_Score"] >= 3:
        return "Champions"
    elif row["R_Score"] >= 3 and row["F_Score"] >= 2:
        return "Loyal Customers"
    elif row["R_Score"] >= 3 and row["F_Score"] == 1:
        return "Recent Customers"
    elif row["R_Score"] <= 2 and row["F_Score"] >= 3:
        return "At-Risk Customers"
    else:
        return "Regular Customers"

rfm["Customer_Segment"] = rfm.apply(
    classify_customer,
    axis=1
)

display(rfm.head(10))
````

````python
# Count customers by segment

segment_summary = (
    rfm.groupby("Customer_Segment")
    .size()
    .reset_index(name="Customer_Count")
    .sort_values("Customer_Count", ascending=False)
)

display(segment_summary)
````

````python
# Plot the customer segment distribution

plt.figure(figsize=(10, 6))

plt.bar(
    segment_summary["Customer_Segment"],
    segment_summary["Customer_Count"]
)

plt.xlabel("Customer Segment")
plt.ylabel("Number of Customers")
plt.title("Customer Segmentation Using RFM Analysis")
plt.xticks(rotation=30, ha="right")

plt.tight_layout()
plt.show()
````
